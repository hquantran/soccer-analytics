Visualization

```python
# Set global visual style
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Distribution of Minutes Played
sns.histplot(df["minutes"], bins=30, kde=True, ax=axes[0, 0], color="#1f77b4")
axes[0, 0].set_title("1. Distribution of Minutes Played")
axes[0, 0].set_xlabel("Minutes")

# 2. Player Count by Position
sns.countplot(
    data=df, 
    y="position", 
    order=df["position"].value_counts().index, 
    ax=axes[0, 1], 
    palette="viridis"
)
axes[0, 1].set_title("2. Player Count by Position")

# 3. Missing Value Percentage
missing_pct = (df.isnull().sum() / len(df)) * 100
missing_pct = missing_pct[missing_pct > 0].sort_values()
missing_pct.plot(kind="barh", ax=axes[1, 0], color="#d62728")
axes[1, 0].set_title("3. Missing Values (%) per Feature")

# 4. Outlier Detection: Goals vs Shots
sns.scatterplot(
    data=df, 
    x="shots_total", 
    y="goals", 
    hue="position", 
    alpha=0.6, 
    ax=axes[1, 1]
)
axes[1, 1].plot([0, 100], [0, 100], "--", color="gray", label="100% Conversion Line")
axes[1, 1].set_title("4. Goals vs. Total Shots (Outlier Spotting)")

plt.tight_layout()
plt.show()
```
